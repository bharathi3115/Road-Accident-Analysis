```python
def categorize_hour(hour):
    if pd.isna(hour):
        return "Unknown"
    elif hour < 6:
        return "Night"
    elif hour < 12:
        return "Morning"
    elif hour < 17:
        return "Afternoon"
    else:
        return "Evening"

df_project["Time_Period"] = df_project["Accident_Hour"].apply(categorize_hour)
```
